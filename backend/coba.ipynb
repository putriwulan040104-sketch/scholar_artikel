```python
import pandas as pd
import scipy.sparse as sp
import os

# Path yang benar - sesuai struktur folder
base_dir     = r"D:\Tugas Akhir\paperci_artikel\backend"
csv_path     = os.path.join(base_dir, 'data', 'cleaned_papers.csv')
tfidf_path   = os.path.join(base_dir, 'data', 'tfidf', 'tfidf_matrix.npz')

print(f'CSV exists    : {os.path.exists(csv_path)}')
print(f'Matrix exists : {os.path.exists(tfidf_path)}')

df           = pd.read_csv(csv_path)
tfidf_matrix = sp.load_npz(tfidf_path)

print(f'Jumlah baris CSV     : {len(df)}')
print(f'Jumlah baris matriks : {tfidf_matrix.shape[0]}')
print(f'Match                : {len(df) == tfidf_matrix.shape[0]}')
```
